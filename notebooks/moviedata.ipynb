```python
from dotenv import load_dotenv
import os
import pandas as pd
from sqlalchemy import create_engine, URL

load_dotenv("D:/books/tmdb-movie-analysis/.env")
password = os.getenv("DB_PASSWORD")

url = URL.create(
    "postgresql+psycopg2",
    username="postgres",
    password=password,
    host="localhost",
    port=5432,
    database="tmdb_project"
)

engine = create_engine(url)

movies = pd.read_csv('D:/books/archive/tmdb_5000_movies.csv')
credits = pd.read_csv('D:/books/archive/tmdb_5000_credits.csv')
credits = credits.rename(columns={'cast': 'cast_list'})

movies.to_sql('movies', engine, if_exists='replace', index=False)
credits.to_sql('credits', engine, if_exists='replace', index=False)

print("Done loading!")
```
